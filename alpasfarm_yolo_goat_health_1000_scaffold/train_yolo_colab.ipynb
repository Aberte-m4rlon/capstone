# 🐐 AlpasFarm — YOLOv8 Goat Health & Scanner 1-Click Trainer
This notebook automatically sets up free GPU acceleration, prepares the 1,000-sample goat dataset, trains YOLOv8 for 50 epochs, and downloads the trained `best.pt` weights for AlpasFarm.

```python
# 1. Install ultralytics
!pip install -q ultralytics
```

```python
# 2. Clone or check GPU
import torch
print("GPU Available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("Device Name:", torch.cuda.get_device_name(0))
```

```python
# 3. Run dataset builder and validation
!python prepare_and_augment_dataset.py
!python validate_yolo_dataset.py
```

```python
# 4. Train YOLOv8 on data.yaml with GPU
from ultralytics import YOLO

model = YOLO('yolov8n.pt')
results = model.train(
    data='data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    save=True,
    project='runs',
    name='goat_health_train'
)
```

```python
# 5. Evaluate validation metrics
metrics = model.val()
print(f"mAP@50: {metrics.box.map50:.4f}")
print(f"Precision: {metrics.box.mp:.4f}")
print(f"Recall: {metrics.box.mr:.4f}")
```

```python
# 6. Export to ONNX and Download best.pt
model.export(format='onnx', imgsz=640)

from google.colab import files
files.download('runs/goat_health_train/weights/best.pt')
files.download('runs/goat_health_train/weights/best.onnx')
```
